```python
import pandas as pd
import speech_recognition as sr
import random
```

```python
# Load the dataset
df = pd.read_csv("C:\\Users\\E1IN\\Downloads\\archive\\spotify-2023.csv", encoding='ISO-8859-1')
print("Loaded!")
df.head()
```

```python
df['mode'] = df['mode'].astype(str).str.strip().str.lower()
df.head()
```

```python
mood_to_mode = {
    "happy": "major", 
    "excited": "major", 
    "joyful": "major", 
    "energetic": "major", 
    "romantic": "major", 
    "sad": "minor", 
    "emotional": "minor", 
    "calm": "minor", 
    "melancholy": "minor", 
    "relaxed": "minor" 
}

print(df[['track_name', 'mode']].head())
```

```python
import speech_recognition as sr


def capture_voice_input(device_index=2):
    r = sr.Recognizer()
    try:
        with sr.Microphone(device_index=device_index) as source:
            print("Speak...")
            r.adjust_for_ambient_noise(source, duration=1)
            print("..........")
            audio = r.listen(source, timeout=10, phrase_time_limit=6)

        try:
            text = r.recognize_google(audio)
            print(f"You said: {text}")
            return text.lower() 
        except sr.UnknownValueError:
            print("Could not understand your voice.")
            return None
        except sr.RequestError as e:
            print(f"Google API error: {e}")
            return None

    except Exception as e:
        print(f" Microphone error: {e}")
        return None
```

```python
def extract_mood_from_text(user_input, mood_keywords):
    for mood in mood_keywords:
        if mood in user_input:
            return mood
    return None
```

```python
def recommend_songs_by_sentence(user_input, df, mood_to_mode):
    mood = extract_mood_from_text(user_input, mood_to_mode.keys())
    if mood:
        target_mode = mood_to_mode[mood]
        matched_songs = df[df['mode'] == target_mode]['track_name'].dropna().unique()
        if len(matched_songs) == 0:
            return [f"No songs found for mood '{mood}'."]
        return random.sample(list(matched_songs), min(5, len(matched_songs)))
    else:
        return ["Sorry, couldn't detect a clear mood."]
```

```python
mic_index = 2  

user_input = capture_voice_input(device_index=mic_index)
# print(f"Final user_input = '{user_input}'")

if user_input is not None and user_input.strip() != "":
    recommendations = recommend_songs_by_sentence(user_input, df, mood_to_mode)
    print("\nRecommended Songs Based on Mood:")
    for song in recommendations:
        print("-", song)
else:
    print("No voice input detected or could not understand what you said.")
```
